## 5. Análisis de granularidad

```python
id_unicos=df["Order ID"].nunique()
numero_registros = df.shape[0]
print(f' El numero de Ordenes unicas es {id_unicos}, menor al numero de registros o filas {numero_registros}')
```

```python
df.groupby("Order ID")["Order Date"].nunique().value_counts()
```

👉 comentario:

Cada orden puede tener múltiples productos, lo que indica que el dataset está a nivel de línea de pedido.

### 🔹 6. Análisis de fechas

```python
# Validar fechas 
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%m/%d/%Y")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%m/%d/%Y")

(df["Ship Date"] < df["Order Date"]).sum()
```

👉 comentario:

No se encontraron inconsistencias temporales
